```python
import pandas as pd
import matplotlib.pyplot as plt 
try:
    df = pd.read_csv('')
    print("CSV loaded successfully.")
except FileNotFoundError:
    print("File not found. Creating a dummy DataFrame for demonstration.")
    
    data = {
        'Product': ['Laptop', 'Mouse', 'Laptop', 'Keyboard', 'Mouse', 'Laptop', 'Keyboard', 'Mouse'],
        'Region': ['North', 'North', 'South', 'South', 'North', 'East', 'East', 'South'],
        'Sales': [1200, 25, 1500, 75, 30, 1100, 60, 45]
    }
    df = pd.DataFrame(data)


print("\n--- Initial DataFrame Head ---")
print(df.head())

print("\n--- Data Information ---")
df.info()

product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)

print("\n--- Total Sales by Product (Data Insight) ---")
print(product_sales)
highest_seller = product_sales.index[0]
highest_sales = product_sales.iloc[0]
print(f"\n Basic Insight 1: The highest selling product is '{highest_seller}' with total sales of ${highest_sales:,}.")

plt.figure(figsize=(10, 6))


product_sales.plot(
    kind='bar',
    color=['skyblue', 'lightcoral', 'lightgreen'], # Customize colors
    title='Total Sales by Product',
    ylabel='Total Sales ($)',
    xlabel='Product'
)


plt.grid(axis='y', linestyle='--')


plt.xticks(rotation=45, ha='right')


plt.tight_layout()


plt.show() 


print("\n Basic Insight 2: The bar plot visually confirms the ranking of products by total sales.")
```
